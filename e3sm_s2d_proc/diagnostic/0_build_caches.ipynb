# RTENSO2026: build the shared caches

Builds, ahead of the other notebooks, the caches they all read: the Reanalysis (BruteForce)
1981-2010 lead-time climatologies for every reference start month (`rc.REFERENCE`, May and Nov)
and the observed anomalies of each field in `FIELDS`. Each is written once under `S2D_DIAG_ROOT`
and reused by the notebooks (`derivation_mode = "auto"`), so they stay fast. A first build takes
about 4 minutes per field and start month (the U850 reads are the slowest); a later run in
`"auto"` mode only checks the caches and rebuilds stale ones (e.g. after an observation product
is extended).

The work is done by `rc.build_caches` in `util/rtenso_case.py`. To run it without opening Jupyter:

```bash
jupyter nbconvert --to notebook --execute --inplace 0_build_caches.ipynb
```

````python
%load_ext autoreload
%autoreload 2

import pandas as pd
from IPython.display import display

from util import rtenso_case as rc
````

### Run controls

````python
# "auto": reuse valid caches and rebuild stale ones; "rebuild": recompute every cache
derivation_mode = "auto"
# Fields to build: lead-time climatologies for all, observed anomalies where rc.FIELDS has a product
FIELDS = rc.CACHE_FIELDS           # ["TS", "PRECT", "TREFHT", "PSL", "U850"]
````

### Build

````python
%%time
rows = rc.build_caches(FIELDS, mode=derivation_mode)
summary = pd.DataFrame(rows)
display(summary.pivot_table(index="field", columns="cache", values="seconds", sort=False))
print(f"total {summary['seconds'].sum() / 60:.1f} min")
````
