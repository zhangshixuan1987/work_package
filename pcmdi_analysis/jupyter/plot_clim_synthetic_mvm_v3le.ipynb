```python
CLIM_MVM_V3LE_REF_DATASET = dataset_for_merged_group(mip=CLIM_MVM_V3LE_MIP, group=CLIM_MVM_V3LE_REF_MERGED_GROUP, case_id=CLIM_MVM_V3LE_CASE_ID, root=CLIM_MVM_V3LE_METRICS_DATA_ROOT)
CLIM_MVM_V3LE_TEST_DATASET = dataset_for_merged_group(mip=CLIM_MVM_V3LE_MIP, group=CLIM_MVM_V3LE_TEST_MERGED_GROUP, case_id=CLIM_MVM_V3LE_CASE_ID, root=CLIM_MVM_V3LE_METRICS_DATA_ROOT)

clim_mvm_v3le_parameters = make_comparison_parameters(
    case_id=CLIM_MVM_V3LE_CASE_ID, ref_group=CLIM_MVM_V3LE_REF_GROUP_LABEL, test_group=CLIM_MVM_V3LE_TEST_GROUP_LABEL,
    ref_dataset=CLIM_MVM_V3LE_REF_DATASET, test_dataset=CLIM_MVM_V3LE_TEST_DATASET,
    metrics_root=CLIM_MVM_V3LE_METRICS_DATA_ROOT, out_dir=CLIM_MVM_V3LE_FIG_DIR,
    test_combined=True, test_model_only=False, show_mean_columns=CLIM_MVM_V3LE_SHOW_MEAN_COLUMNS,
    clim_vars=CLIM_MVM_V3LE_CLIM_VARS, clim_regions=CLIM_MVM_V3LE_CLIM_REGIONS, error_norm=CLIM_MVM_V3LE_ERROR_NORM,
    figure_format=CLIM_MVM_V3LE_FIGURE_FORMAT, clim_portrait_figure_size=CLIM_MVM_V3LE_PORTRAIT_FIGURE_SIZE,
    clim_parcoord_figure_size=CLIM_MVM_V3LE_PARCOORD_FIGURE_SIZE,
    clim_viewer=True, mova_viewer=False, movc_viewer=False, enso_viewer=False,
)

if CLIM_MVM_V3LE_RUN_PLOTS:
    run_synthetic_plots(clim_mvm_v3le_parameters, title_label=CLIM_MVM_V3LE_TITLE_LABEL, metric_file=CLIM_MVM_V3LE_METRIC_CONFIG_FILE, metric_selection={"mean_climate": CLIM_MVM_V3LE_METRICS}, show_figures=CLIM_MVM_V3LE_SHOW_FIGURES, debug=CLIM_MVM_V3LE_DEBUG)
else:
    print("Set CLIM_MVM_V3LE_RUN_PLOTS = True to generate figures.")
```
