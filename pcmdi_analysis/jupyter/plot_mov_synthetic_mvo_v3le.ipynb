```python
# Variability-mode synthetic metrics: v3.LE historical ensemble versus CMIP.
MOV_MVO_V3LE_RUN_PLOTS = True
MOV_MVO_V3LE_SHOW_FIGURES = True
MOV_MVO_V3LE_DEBUG = True
MOV_MVO_V3LE_CASE_ID = "v20260212"
MOV_MVO_V3LE_TITLE_LABEL = "E3SMv3 Historical versus CMIP"
MOV_MVO_V3LE_FIGURE_FORMAT = "pdf"
MOV_MVO_V3LE_FIGURE_SIZE = (80.0, 30.0)  # Backward-compatible fallback.
MOV_MVO_V3LE_PORTRAIT_FIGURE_SIZE = (60.0, 25.0)
MOV_MVO_V3LE_PARCOORD_FIGURE_SIZE = (70.0, 30.0)
MOV_MVO_V3LE_PARCOORD_LEGEND_Y_OFFSET = -0.28  # Clear rotated mode labels.

MOV_MVO_V3LE_METRICS_DATA_ROOT = Path(
    "/lcrc/group/e3sm/diagnostics/pcmdi_data/metrics_data"
)
MOV_MVO_V3LE_METRIC_CONFIG_FILE = PROJECT_ROOT / "config" / "synthetic_metrics_list.json"
MOV_MVO_V3LE_FIG_DIR = Path(
    "/lcrc/group/e3sm/public_html/diagnostic_output/ac.szhang/"
    "e3sm-pcmdi-le/paper_plot"
)

# Existing merged 25-member v3.LE historical dataset and CMIP reference.
MOV_MVO_V3LE_TEST_MIP = "e3sm"
MOV_MVO_V3LE_TEST_MERGED_GROUP = "historical"
MOV_MVO_V3LE_TEST_GROUP_LABEL = "E3SMv3-Historical"
MOV_MVO_V3LE_REF_GROUP_LABEL = "CMIP"
MOV_MVO_V3LE_ALIGN_WITH_CMIP = True
MOV_MVO_V3LE_TEST_COMBINED = True
MOV_MVO_V3LE_TEST_MODEL_ONLY = False

# Explicit metrics and expected figure types.
MOV_MVO_V3LE_METRICS = (
    "rms",                   # RMSE; portrait and parallel-coordinate figures
    "rmsc",                  # Centered RMSE; portrait and parallel-coordinate figures
    "stdv_pc_ratio_to_obs",  # PC standard-deviation ratio; both figure types
)

# Explicit figure-set selection: exactly this diagnostic family is enabled.
MOV_MVO_V3LE_CLIM_VIEWER = False
MOV_MVO_V3LE_MOVA_VIEWER = True
MOV_MVO_V3LE_MOVC_VIEWER = True
MOV_MVO_V3LE_ENSO_VIEWER = False

MOV_MVO_V3LE_SHOW_MEAN_COLUMNS = True
MOV_MVO_V3LE_MOVS_GROUP = "cbf"
MOV_MVO_V3LE_ATM_MODES = ("NAM", "NAO", "PNA", "NPO")
MOV_MVO_V3LE_ATM_OBS = "NOAA-20C"
MOV_MVO_V3LE_CPL_MODES = ("PDO", "NPGO")
MOV_MVO_V3LE_CPL_OBS = "HadISST"
MOV_MVO_V3LE_ERROR_NORM = "reference"
MOV_MVO_V3LE_EXCLUDE_MODELS = [
    "E3SM-1-0", "E3SM-1-1", "E3SM-1-1-ECA", "E3SM-2-0", "E3SM-2-1"
]
```
