```python
ENSO_MVO_V3LE_TEST_DATASET = dataset_for_merged_group(
    mip=ENSO_MVO_V3LE_TEST_MIP,
    group=ENSO_MVO_V3LE_TEST_MERGED_GROUP,
    case_id=ENSO_MVO_V3LE_CASE_ID,
    root=ENSO_MVO_V3LE_METRICS_DATA_ROOT,
)

enso_mvo_v3le_parameters = make_comparison_parameters(
    case_id=ENSO_MVO_V3LE_CASE_ID,
    ref_group=ENSO_MVO_V3LE_REF_GROUP_LABEL,
    test_group=ENSO_MVO_V3LE_TEST_GROUP_LABEL,
    test_dataset=ENSO_MVO_V3LE_TEST_DATASET,
    metrics_root=ENSO_MVO_V3LE_METRICS_DATA_ROOT,
    out_dir=ENSO_MVO_V3LE_FIG_DIR,
    align_with_cmip=ENSO_MVO_V3LE_ALIGN_WITH_CMIP,
    test_combined=ENSO_MVO_V3LE_TEST_COMBINED,
    test_model_only=ENSO_MVO_V3LE_TEST_MODEL_ONLY,
    show_mean_columns=ENSO_MVO_V3LE_SHOW_MEAN_COLUMNS,
    exclude_models=ENSO_MVO_V3LE_EXCLUDE_MODELS,
    figure_format=ENSO_MVO_V3LE_FIGURE_FORMAT,
    enso_figure_size=ENSO_MVO_V3LE_FIGURE_SIZE,
    enso_reduced_set=ENSO_MVO_V3LE_REDUCED_SET,
    enso_metric_order=ENSO_MVO_V3LE_METRIC_ORDER,
    enso_model_order=ENSO_MVO_V3LE_MODEL_ORDER,
    enso_sort_y_names=ENSO_MVO_V3LE_SORT_Y_NAMES,
    enso_show_project_means=ENSO_MVO_V3LE_SHOW_PROJECT_MEANS,
    enso_show_ref_row=ENSO_MVO_V3LE_SHOW_REF_ROW,
    enso_show_alt_obs_rows=ENSO_MVO_V3LE_SHOW_ALT_OBS_ROWS,
    clim_viewer=ENSO_MVO_V3LE_CLIM_VIEWER,
    mova_viewer=ENSO_MVO_V3LE_MOVA_VIEWER,
    movc_viewer=ENSO_MVO_V3LE_MOVC_VIEWER,
    enso_viewer=ENSO_MVO_V3LE_ENSO_VIEWER,
)

if ENSO_MVO_V3LE_RUN_PLOTS:
    run_synthetic_plots(
        enso_mvo_v3le_parameters,
        title_label=ENSO_MVO_V3LE_TITLE_LABEL,
        metric_file=ENSO_MVO_V3LE_METRIC_CONFIG_FILE,
        metric_selection={"enso_metric": ENSO_MVO_V3LE_METRICS},
        show_figures=ENSO_MVO_V3LE_SHOW_FIGURES,
        debug=ENSO_MVO_V3LE_DEBUG,
    )
else:
    print("Set ENSO_MVO_V3LE_RUN_PLOTS = True to generate figures.")
```
