```python
# Mean-climate synthetic metrics: v3.LE historical ensemble versus CMIP.
CLIM_MVO_V3LE_RUN_PLOTS = True
CLIM_MVO_V3LE_SHOW_FIGURES = True
CLIM_MVO_V3LE_DEBUG = True
CLIM_MVO_V3LE_CASE_ID = "v20260212"
CLIM_MVO_V3LE_TITLE_LABEL = "E3SMv3 Historical versus CMIP"
CLIM_MVO_V3LE_FIGURE_FORMAT = "pdf"
CLIM_MVO_V3LE_FIGURE_SIZE = (50.0, 20.0)  # Backward-compatible fallback.
CLIM_MVO_V3LE_PORTRAIT_FIGURE_SIZE = (60.0, 25.0)
CLIM_MVO_V3LE_PARCOORD_FIGURE_SIZE = (70.0, 20.0)  # More column spacing.

CLIM_MVO_V3LE_METRICS_DATA_ROOT = Path(
    "/lcrc/group/e3sm/diagnostics/pcmdi_data/metrics_data"
)
CLIM_MVO_V3LE_METRIC_CONFIG_FILE = PROJECT_ROOT / "config" / "synthetic_metrics_list.json"
CLIM_MVO_V3LE_FIG_DIR = Path(
    "/lcrc/group/e3sm/public_html/diagnostic_output/ac.szhang/"
    "e3sm-pcmdi-le/paper_plot"
)

# Existing merged 25-member v3.LE historical dataset and CMIP reference.
CLIM_MVO_V3LE_TEST_MIP = "e3sm"
CLIM_MVO_V3LE_TEST_MERGED_GROUP = "historical"
CLIM_MVO_V3LE_TEST_GROUP_LABEL = "E3SMv3-Historical"
CLIM_MVO_V3LE_REF_GROUP_LABEL = "CMIP"
CLIM_MVO_V3LE_ALIGN_WITH_CMIP = True
CLIM_MVO_V3LE_TEST_COMBINED = True
CLIM_MVO_V3LE_TEST_MODEL_ONLY = False

# Explicit metrics and expected figure types.
CLIM_MVO_V3LE_METRICS = (
    "mae_xy",   # Seasonal mean absolute error; portrait figure
    "cor_xy",   # Seasonal spatial pattern correlation; portrait figure
    "rms_xy",   # Seasonal spatial RMSE; portrait figure
    "bias_xy",  # Annual mean bias; parallel-coordinate figure
    "rms_xyt",  # Annual spatiotemporal RMSE; parallel-coordinate figure
)

# Explicit figure-set selection: exactly this diagnostic family is enabled.
CLIM_MVO_V3LE_CLIM_VIEWER = True
CLIM_MVO_V3LE_MOVA_VIEWER = False
CLIM_MVO_V3LE_MOVC_VIEWER = False
CLIM_MVO_V3LE_ENSO_VIEWER = False

CLIM_MVO_V3LE_SHOW_MEAN_COLUMNS = True
CLIM_MVO_V3LE_CLIM_VARS = (
    "pr,prw,psl,rlds,rldscs,rltcre,rstcre,rsus,rsuscs,rlus,rlut,rlutcs,"
    "rsds,rsdscs,rsut,rsutcs,rtmt,sfcWind,tas,tauu,tauv,ts,ta-200,ta-850,"
    "ua-200,ua-850,va-200,va-850,zg-500"
)
CLIM_MVO_V3LE_CLIM_REGIONS = "global"
CLIM_MVO_V3LE_ERROR_NORM = "reference"
CLIM_MVO_V3LE_EXCLUDE_VARS = {
    "E3SM-1-0": ["ta-850"],
    "E3SM-1-1-ECA": ["ta-850"],
    "CIESM": ["pr"],
    "KIOST-ESM": ["zg-500", "ta-850"],
    "GISS-E2-2-G": ["rlutcs", "zg-500"],
}
CLIM_MVO_V3LE_EXCLUDE_MODELS = [
    "E3SM-1-0", "E3SM-1-1", "E3SM-1-1-ECA", "E3SM-2-0", "E3SM-2-1"
]
```

```python
CLIM_MVO_V3LE_TEST_DATASET = dataset_for_merged_group(
    mip=CLIM_MVO_V3LE_TEST_MIP,
    group=CLIM_MVO_V3LE_TEST_MERGED_GROUP,
    case_id=CLIM_MVO_V3LE_CASE_ID,
    root=CLIM_MVO_V3LE_METRICS_DATA_ROOT,
)

clim_mvo_v3le_parameters = make_comparison_parameters(
    case_id=CLIM_MVO_V3LE_CASE_ID,
    ref_group=CLIM_MVO_V3LE_REF_GROUP_LABEL,
    test_group=CLIM_MVO_V3LE_TEST_GROUP_LABEL,
    test_dataset=CLIM_MVO_V3LE_TEST_DATASET,
    metrics_root=CLIM_MVO_V3LE_METRICS_DATA_ROOT,
    out_dir=CLIM_MVO_V3LE_FIG_DIR,
    align_with_cmip=CLIM_MVO_V3LE_ALIGN_WITH_CMIP,
    test_combined=CLIM_MVO_V3LE_TEST_COMBINED,
    test_model_only=CLIM_MVO_V3LE_TEST_MODEL_ONLY,
    show_mean_columns=CLIM_MVO_V3LE_SHOW_MEAN_COLUMNS,
    clim_vars=CLIM_MVO_V3LE_CLIM_VARS,
    clim_regions=CLIM_MVO_V3LE_CLIM_REGIONS,
    error_norm=CLIM_MVO_V3LE_ERROR_NORM,
    exclude_vars=CLIM_MVO_V3LE_EXCLUDE_VARS,
    exclude_models=CLIM_MVO_V3LE_EXCLUDE_MODELS,
    figure_format=CLIM_MVO_V3LE_FIGURE_FORMAT,
    clim_figure_size=CLIM_MVO_V3LE_FIGURE_SIZE,
    clim_portrait_figure_size=CLIM_MVO_V3LE_PORTRAIT_FIGURE_SIZE,
    clim_parcoord_figure_size=CLIM_MVO_V3LE_PARCOORD_FIGURE_SIZE,
    clim_viewer=CLIM_MVO_V3LE_CLIM_VIEWER,
    mova_viewer=CLIM_MVO_V3LE_MOVA_VIEWER,
    movc_viewer=CLIM_MVO_V3LE_MOVC_VIEWER,
    enso_viewer=CLIM_MVO_V3LE_ENSO_VIEWER,
)

if CLIM_MVO_V3LE_RUN_PLOTS:
    run_synthetic_plots(
        clim_mvo_v3le_parameters,
        title_label=CLIM_MVO_V3LE_TITLE_LABEL,
        metric_file=CLIM_MVO_V3LE_METRIC_CONFIG_FILE,
        metric_selection={"mean_climate": CLIM_MVO_V3LE_METRICS},
        show_figures=CLIM_MVO_V3LE_SHOW_FIGURES,
        debug=CLIM_MVO_V3LE_DEBUG,
    )
else:
    print("Set CLIM_MVO_V3LE_RUN_PLOTS = True to generate figures.")
```
