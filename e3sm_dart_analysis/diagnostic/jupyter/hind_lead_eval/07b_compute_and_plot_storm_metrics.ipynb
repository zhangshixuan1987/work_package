# 07b. Case study: the January 2012 Pacific Northwest storm: member storm metrics

## Metrics (every member, start and region)

On the event window: event-total and maximum daily precipitation, peak-precipitation date error (searched within +-`peak_tolerance_days`), event-mean and minimum daily-mean temperature, and **cold-wet coincidence**: precipitation on days with daily-mean temperature at or below 0 C and +1 C (not diagnosed snowfall or freezing rain). Spatial: pattern correlation and RMSE over the two regions of event-total precipitation and event-mean temperature. Temperature is used raw (`none`) and with the model's mean bias removed (`amip`: minus the AMIP-minus-observed climatology of the day), so the freezing counts are not set by the model's climate offset (`biases`).

Design and event definition: `07a`.

**The 07a-07d series** (January 2012 Pacific Northwest storm; each notebook reads the data it needs, computes, saves, then plots): `07a` regional series; `07b` member storm metrics; `07c` event maps; `07d` storm environment. Settings: `WinterStormSettings` in `configs/hind_lead_eval_cases.py`; products: `util/winter_storm_workflow.py`; methods: `util/winter_storm_case.py`. Design and event definition: `07a`.

Every product is saved with the settings it was computed from and read back on the next run when those settings are unchanged; `FORCE_COMPUTE = True` recomputes this notebook's results, `FORCE_INPUTS = True` also re-reads the model output and observations. Model output is read only when a product has to be (re)computed.

## Outputs

`data/hind_lead_eval/winter_storm_2012/`: `storm_metrics.csv` (members), `storm_metrics_obs.csv`; figures `storm_metrics_<REGION>_<BIAS>.png`.
